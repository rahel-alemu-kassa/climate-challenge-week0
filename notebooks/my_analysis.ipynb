```python
import pandas as pd

ethiopia_data = pd.read_csv('ethiopia.csv')

ethiopia_data.head()
```

```python
ethiopia_data.info()
ethiopia_data.isnull().sum()
```

```python
import matplotlib.pyplot as plt

ethiopia_data.groupby('YEAR')['T2M'].mean().plot(kind='line')
plt.title('Average Temperature in Ethiopia (2015-2026)')
plt.show()
```

```python

print("Missing values in each column:")
print(ethiopia_data.isnull().sum())


print("\nNumber of duplicate rows:", ethiopia_data.duplicated().sum())
```

```python
ethiopia_data.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.lineplot(data=ethiopia_data, x='YEAR', y='T2M', marker='o', color='red')

plt.title('Temperature Trend in Ethiopia (2015-2026)', fontsize=15)
plt.xlabel('Year')
plt.ylabel('Average Temp (°C)')
plt.grid(True)
plt.show()
```

```python

plt.figure(figsize=(10, 6))
ethiopia_data.groupby('YEAR')['PRECTOTCORR'].sum().plot(kind='bar', color='skyblue')

plt.title('Total Yearly Rainfall in Ethiopia (2015-2026)', fontsize=15)
plt.xlabel('Year')
plt.ylabel('Total Rainfall (mm)')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

files = ['ethiopia.csv', 'kenya.csv', 'nigeria.csv', 'sudan.csv', 'tanzania.csv']
countries = ['Ethiopia', 'Kenya', 'Nigeria', 'Sudan', 'Tanzania']

fig, axes = plt.subplots(nrows=5, ncols=1, figsize=(10, 20))

for i, file in enumerate(files):
    df = pd.read_csv(file)
    
    yearly_temp = df.groupby('YEAR')['T2M'].mean()
    
    yearly_temp.plot(ax=axes[i], marker='s', color='orange')
    axes[i].set_title(f'Temperature Trend: {countries[i]}')
    axes[i].set_ylabel('Temp (°C)')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

eth = pd.read_csv('ethiopia.csv').groupby('YEAR')['T2M'].mean()
sud = pd.read_csv('sudan.csv').groupby('YEAR')['T2M'].mean()

plt.plot(eth.index, eth.values, label='Ethiopia', color='blue', marker='o')
plt.plot(sud.index, sud.values, label='Sudan', color='red', marker='x')

plt.title('Temperature Comparison: Ethiopia vs Sudan')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Find the row with the highest temperature in Ethiopia
hottest_day = ethiopia_data.loc[ethiopia_data['T2M'].idxmax()]

print(f"The hottest day was in the year {hottest_day['YEAR']}, on day {hottest_day['DOY']}.")
print(f"The temperature reached {hottest_day['T2M']}°C.")
```
